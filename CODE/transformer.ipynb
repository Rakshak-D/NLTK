```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents=[
    "Machine learning",
    "Deep learning ",
    "Natural language processing"
]
```

```python
vectorizer=CountVectorizer(stop_words="english",ngram_range=(1,2),max_features=20)
```

```python
x=vectorizer.fit_transform(documents)
```

```python
feature_names=vectorizer.get_feature_names_out()
```

```python
print(feature_names)
```

```python
print(x.toarray())
```

```python
from gensim.models import Word2Vec
```

```python
sentences=[
    ["natural","language","processing"],
    ["word","embeddings","are","vectors"],
    ["nlp","is","a","filed","of","ai"]
]
```

```python
model=Word2Vec(sentences,vector_size=10,window=5,min_count=1,workers=4)
```

```python
vector=model.wv['nlp']
```

```python
vector
```

```python
model.wv.most_similar("nlp",topn=2)
```
